```python
import pandas as pd
import numpy as np

df = pd.read_csv("SuperMarket Analysis.csv")

print("Dataset loaded successfully!")
```

```python
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nColumn Names:")
print(df.columns.tolist())
```

```python
print("Missing Values:")
print(df.isnull().sum())
```

```python
duplicates = df.duplicated().sum()

print("Number of duplicate rows:", duplicates)
```

```python
df.columns = [
    'Invoice_ID',
    'Branch',
    'City',
    'Customer_Type',
    'Gender',
    'Product_Line',
    'Unit_Price',
    'Quantity',
    'Tax_5',
    'Sales',
    'Date',
    'Time',
    'Payment',
    'COGS',
    'Gross_Margin_Percentage',
    'Gross_Income',
    'Rating'
]

print(df.columns.tolist())
```

```python
text_columns = [
    'Invoice_ID',
    'Branch',
    'City',
    'Customer_Type',
    'Gender',
    'Product_Line',
    'Payment'
]

for col in text_columns:
    df[col] = df[col].str.strip()

print("Text data cleaned successfully!")
```

```python
for col in [
    'Branch',
    'City',
    'Customer_Type',
    'Gender',
    'Product_Line',
    'Payment'
]:
    print("\n", col)
    print(df[col].unique())
```

```python
df['Date'] = pd.to_datetime(
    df['Date'],
    errors='coerce'
)

df['Time'] = pd.to_datetime(
    df['Time'],
    format='%H:%M:%S',
    errors='coerce'
).dt.time

print(df[['Date', 'Time']].head())
```

```python
print(df['Date'].dtype)
```

```python
df['Year'] = df['Date'].dt.year

df['Month'] = df['Date'].dt.month

df['Month_Name'] = df['Date'].dt.month_name()

df['Quarter'] = df['Date'].dt.quarter

df['Day'] = df['Date'].dt.day

df['Day_Name'] = df['Date'].dt.day_name()

df['Day_Of_Week'] = df['Date'].dt.dayofweek + 1
```

```python
df[
    [
        'Date',
        'Year',
        'Month',
        'Month_Name',
        'Quarter',
        'Day',
        'Day_Name',
        'Day_Of_Week'
    ]
].head()
```

```python
df['Hour'] = df['Time'].apply(
    lambda x: x.hour if pd.notnull(x) else None
)
```

```python
def get_time_period(hour):

    if pd.isnull(hour):
        return 'Unknown'

    elif hour < 12:
        return 'Morning'

    elif hour < 17:
        return 'Afternoon'

    elif hour < 21:
        return 'Evening'

    else:
        return 'Night'


df['Time_Period'] = df['Hour'].apply(get_time_period)
```

```python
df[['Time', 'Hour', 'Time_Period']].head(10)
```

```python
df['Calculated_COGS'] = (
    df['Unit_Price'] * df['Quantity']
)

df['Calculated_Tax'] = (
    df['Calculated_COGS'] * 0.05
)

df['Calculated_Gross_Profit'] = (
    df['Sales'] - df['Calculated_COGS']
)
```

```python
df[
    [
        'Unit_Price',
        'Quantity',
        'Sales',
        'Calculated_COGS',
        'Calculated_Tax',
        'Calculated_Gross_Profit'
    ]
].head()
```

```python
valid_sales = df['Sales'] > 0

valid_quantity = df['Quantity'] > 0

valid_price = df['Unit_Price'] > 0

valid_rating = df['Rating'].between(1, 10)

all_valid = (
    valid_sales &
    valid_quantity &
    valid_price &
    valid_rating
)
```

```python
df_valid = df[all_valid].copy()

df_invalid = df[~all_valid].copy()

print("Total Records:", len(df))
print("Valid Records:", len(df_valid))
print("Invalid Records:", len(df_invalid))
```

```python
if df_invalid.empty:
    print("All records passed validation!")
else:
    print("Invalid records found:")
    display(df_invalid.head())
```

```python
dim_date = df_valid[
    [
        'Date',
        'Year',
        'Month',
        'Month_Name',
        'Quarter',
        'Day',
        'Day_Name',
        'Day_Of_Week'
    ]
].drop_duplicates().copy()

dim_date = dim_date.sort_values('Date').reset_index(drop=True)

dim_date.insert(
    0,
    'Date_Key',
    range(1, len(dim_date) + 1)
)

display(dim_date.head())
```

```python
dim_branch = df_valid[
    ['Branch', 'City']
].drop_duplicates().reset_index(drop=True)

dim_branch.insert(
    0,
    'Branch_Key',
    range(1, len(dim_branch) + 1)
)

display(dim_branch)
```

```python
dim_product = df_valid[
    ['Product_Line']
].drop_duplicates().reset_index(drop=True)

dim_product.insert(
    0,
    'Product_Key',
    range(1, len(dim_product) + 1)
)

display(dim_product)
```

```python
dim_customer = df_valid[
    [
        'Customer_Type',
        'Gender'
    ]
].drop_duplicates().reset_index(drop=True)

dim_customer.insert(
    0,
    'Customer_Key',
    range(1, len(dim_customer) + 1)
)

display(dim_customer)
```

```python
dim_payment = df_valid[
    ['Payment']
].drop_duplicates().reset_index(drop=True)

dim_payment.insert(
    0,
    'Payment_Key',
    range(1, len(dim_payment) + 1)
)

display(dim_payment)
```

```python
fact_sales = df_valid.copy()
```

```python
fact_sales = fact_sales.merge(
    dim_date[['Date_Key', 'Date']],
    on='Date',
    how='left'
)
```

```python
fact_sales = fact_sales.merge(
    dim_branch,
    on=['Branch', 'City'],
    how='left'
)
```

```python
fact_sales = fact_sales.merge(
    dim_product,
    on='Product_Line',
    how='left'
)
```

```python
fact_sales = fact_sales.merge(
    dim_customer,
    on=['Customer_Type', 'Gender'],
    how='left'
)
```

```python
fact_sales = fact_sales.merge(
    dim_payment,
    on='Payment',
    how='left'
)
```

```python
fact_sales = fact_sales[
    [
        'Invoice_ID',
        'Date_Key',
        'Branch_Key',
        'Product_Key',
        'Customer_Key',
        'Payment_Key',
        'Time',
        'Hour',
        'Time_Period',
        'Unit_Price',
        'Quantity',
        'Tax_5',
        'Sales',
        'COGS',
        'Gross_Margin_Percentage',
        'Gross_Income',
        'Rating'
    ]
]

display(fact_sales.head())
```
